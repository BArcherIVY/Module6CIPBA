```python
#15.1
import multiprocessing
import time
import random

def print_time():
    wait_time = random.random()
    time.sleep(wait_time)
    print(time.ctime(), flush=True)

processes = []

for i in range(3):
    p = multiprocessing.Process(target=print_time)
    p.start()
    processes.append(p)

for p in processes:
    p.join()
```
